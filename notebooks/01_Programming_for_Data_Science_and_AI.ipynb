# 4. Operators
Operators let us compute with and compare data.

## Arithmetic operators

````python
a = 11
b = 3

print("a + b  =", a + b)    # addition
print("a - b  =", a - b)    # subtraction
print("a * b  =", a * b)    # multiplication
print("a / b  =", a / b)    # division
print("a // b =", a // b)   # integer (floor) division
print("a % b  =", a % b)    # modulo (remainder)
print("a ** b =", a ** b)   # exponent
````

### Operator precedence
* Order of precedence applies, just like in normal maths (and to *all* operators).
* Full order: [Python docs](https://docs.python.org/3/reference/expressions.html#operator-precedence).

**» Predict** — These two lines are **not** the same. **Write down both numbers you expect**, then run.

````python
# Multiplication has higher precedence than addition
print(1 + 2 * 3)
print((1 + 2) * 3)
````

## Comparison operators
Each comparison produces a **boolean** (`True`/`False`).

````python
age = 5

print("age == 4:", age == 4)   # equal to
print("age != 4:", age != 4)   # not equal to
print("age > 4: ", age > 4)    # greater than
print("age >= 4:", age >= 4)   # greater than or equal to
print("age < 4: ", age < 4)    # less than
print("age <= 4:", age <= 4)   # less than or equal to
````

Note the difference between `==` and `=`:

* A single `=` **assigns** a value (puts a label on a box).
* A double `==` **asks a question** — "are these the same?" — and gives `True`/`False`.

## Strings
Manipulating text is extremely common in data science (codes, notes, labels).

### Accessing characters
Characters in a string are accessed just like list items.

````python
diagnosis = "N18.2 - Chronic kidney disease, stage 2"

# First character
print(diagnosis[0])
# Last character
print(diagnosis[-1])
# A slice of characters
print(diagnosis[8:30])

# 'len' works on strings too
print('Length of string', len(diagnosis))
````

### String methods
* Strings have many helpful [methods](https://docs.python.org/3/library/stdtypes.html#str).
* Strings are **immutable**: methods return a *modified copy*, they do not change the original (unlike list methods). Always check the docs.

````python
# Change case
print('Upper case:', diagnosis.upper())
print('Lower case:', diagnosis.lower())

# Split into a list of pieces at a chosen separator
diagnosis_split = diagnosis.split(' - ')
print('Diagnosis code:', diagnosis_split[0])
print('Diagnosis description:', diagnosis_split[1])

# Replace part of the string
print('Replaced:', diagnosis.replace('Chronic kidney disease', 'CKD'))

# f-strings make printing with variables easy
print(f'The diagnosis is: {diagnosis}')
````

# 6. Type conversion (casting)
* With real data you often need to **convert** values between types.
* Numbers frequently arrive as **text (strings)** — e.g. read from a file or a form — and you cannot do maths on text until you convert it.
* See the [docs](https://docs.python.org/3/library/functions.html) for each conversion. Here we build a small BMI calculation.

````python
# Imagine these arrived as text from a data file
weight_text = "80"    # kg, as a string
height_text = "1.8"   # m, as a string
print(type(weight_text), type(height_text))

# We cannot do maths on strings, so convert (cast) them first
weight = int(weight_text)     # string -> integer
height = float(height_text)   # string -> float

bmi = weight / height ** 2
print('BMI =', round(bmi, 1))
````

# 7. Control flow
Control flow lets your program make decisions and repeat work, instead of running every line once, top to bottom.

## If / elif / else
These let your code do different things depending on a condition:

> if `condition` then do `action`
>
> else do `default action`

You can chain several conditions with `elif`. **Indentation marks which lines belong to each branch — it is not optional in Python.**

A resting heart rate, classified with if / else.

````python
heart_rate = 110   # BPM

# if ... then
if heart_rate > 100:
    print('Heart rate of', heart_rate, 'BPM indicates tachycardia')
````

````python
# if ... then ... else
if heart_rate > 100:
    print('Heart rate of', heart_rate, 'BPM indicates tachycardia')
else:
    print('Heart rate of', heart_rate, 'BPM is normal')
````

Extending it with elif, and a more complex (compound) condition.

````python
heart_rate = 110   # BPM
age = 34           # years

# if ... elif ... else
if heart_rate > 100:
    print('Heart rate of', heart_rate, 'BPM indicates TACHYCARDIA')
elif heart_rate < 60:
    print('Heart rate of', heart_rate, 'BPM indicates BRADYCARDIA')
else:
    print('Heart rate of', heart_rate, 'BPM is normal')
````

````python
# A condition can be any boolean expression
if heart_rate > 100 and age >= 18:
    print('- Indicates ADULT tachycardia (age', age, 'years)')
````

## For loops
A for loop repeats an action for each item in a sequence:

> for each `item` in `sequence` do `action`

Instead of writing `print(procedures[0])`, `print(procedures[1])`, ... we loop.

Print every procedure on the list.

````python
# The lists as booked (the Lists section above sorted and changed them)
procedures = ['Hernia repair', 'Appendicectomy', 'Cholecystectomy', 'Colectomy', 'Mastectomy']
medications = ['Metformin', 'Lisinopril', 'Atorvastatin', 'Amlodipine', 'Furosemide']

# For each 'procedure' in the list 'procedures'
for procedure in procedures:
    print(procedure)
````

### The range function
* `range()` generates a sequence of numbers between limits — useful for loop indices.
* Combined with `len()`, it lets us step through two lists at once by index.

````python
# The booked duration of each procedure, in minutes, in the same order
durations = [60, 45, 90, 180, 120]

n_procedures = len(procedures)

# index takes the values 0, 1, 2, ... up to n_procedures - 1
for index in range(n_procedures):
    print(procedures[index], durations[index], 'min')
````

## While loops
A while loop repeats *while a condition stays true*:

> while `condition` do `action`

Something inside the loop must eventually make the condition false, or it runs forever. Indentation matters here too.

Print each procedure and its duration using a while loop and a manual index.

````python
# Start the index at zero
index = 0

# Keep going while the index is still inside the list
while index < len(durations):
    print(procedures[index], durations[index], 'min')
    index = index + 1   # increment, or the loop never ends
````

### While conditions can be any boolean expression
For example: keep printing only while each case is shorter than two hours.

````python
a = 11
b = 3

a = a + b      # the long way
print("a + b ->", a)

a += b         # shorthand for a = a + b
print("a += b ->", a)
a -= b
print("a -= b ->", a)
a *= b
print("a *= b ->", a)
````

## Additional work 6 — converting to bool
* Any type can be converted to `bool`. This is sometimes used to simplify comparisons.
* Use sparingly — it can harm readability.

````python
print("bool(0)     =", bool(0))      # 0 becomes False
print("bool(1)     =", bool(1))      # any non-zero number becomes True
print("bool(3452)  =", bool(3452))
print("bool('')    =", bool(''))     # empty string becomes False
print("bool('bus') =", bool('bus'))  # any non-empty string becomes True
````

### From low-level to modern high-level languages

**Low-level languages** *(e.g. Assembly, machine code)* — let computer instructions be written in a standardised language, but are very close to the hardware:

```asm
section .data
    message: db "Hello, world!", 10
    len: equ $-message
section .text
    global _start
_start:
    mov eax, 4      ; system call number for write
    mov ebx, 1      ; file descriptor for standard output
    mov ecx, message ; address of the message
    mov edx, len    ; length of the message
    int 0x80        ; invoke the system call
```

**Early high-level languages** *(e.g. FORTRAN, C)* — abstracted from the architecture, improving portability and reusability:

```fortran
program hello_world
print *, "Hello, world!"
end program hello_world
```

**Modern high-level languages** *(e.g. Python, JavaScript)* — more abstraction, more simplicity, more error tolerance, more capability. The same program in Python is a single line:

```python
print("Hello, World!")
```
