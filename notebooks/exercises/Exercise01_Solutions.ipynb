## Operators
### Predict: Precedence — `1 + 2 * 3`
**A common wrong mental model:** reading strictly left to right gives `(1 + 2) * 3 = 9`. **Why that is wrong:** Python (like maths) does `*` *before* `+`, so it is `1 + (2 * 3) = 7`. The second line uses brackets to *force* the addition first, giving `9`. Brackets are the safe way to make your intent explicit.

```python
print(1 + 2 * 3)
print((1 + 2) * 3)
```

### Predict: Comparisons
1. `5 == 5.0` is **`True`**: an `int` and a `float` are compared by value.
2. `'5' == 5` is **`False`**: text never equals a number. There is no error, just `False`, so a value read in as text can quietly fail every check.
3. `'10' < '9'` is **`True`**: text is compared character by character, and `'1'` comes before `'9'`. Numbers stored as text sort in the wrong order; convert them first (see Casting).

```python
print(5 == 5.0)
print('5' == 5)
print('10' < '9')
```

```python
asthma = False
female = True
print(asthma and female)
print(asthma or female)
print(not asthma)
print(female and not asthma)
```

### Predict: Precedence and logic
**1. `2 ** 3` = 8** — `**` is exponent, so 2-cubed.

**2. `9 - 4 // 3` = 8** — *the trap.* A common wrong answer is `(9 - 4) // 3 = 1`. But `//` has higher precedence than `-`, so it is `9 - (4 // 3) = 9 - 1 = 8`.

**3. `not (True and False)` = True** — inside is `False`; `not False` is `True`.

**4. `not ('tree' != 'cat') or (6 < 7) or (4724 / 63 > 75)` = True** — break it down:

* `'tree' != 'cat'` is `True`, so `not (True)` is `False`.
* `6 < 7` is `True`.
* `4724 / 63 ≈ 74.98`, which is **not** `> 75`, so `False`.
* `False or True or False` → **True** (`or` is true if *any* part is true).

```python
print(2 ** 3)
print(9 - 4 // 3)
print(not (True and False))
print(not ('tree' != 'cat') or (6 < 7) or (4724 / 63 > 75))
```

## Casting
### Predict: Which conversions work?
* `int("80")` **works**: `80`.
* `int("7.8")` **errors**: `ValueError: invalid literal for int() with base 10: '7.8'`. `int()` only accepts text that is a whole number. Use `float()` for text with a decimal point.
* `float("7.8")` **works**: `7.8`.
* `int(7.8)` is **surprising**: `7`. Converting a number to `int` *cuts off* the decimal part, it does not round. Use `round()` when you mean rounding; silently truncating a lab value is a real error.

```python
print(int("80"))
print(float("7.8"))
print(int(7.8))
print(round(7.8))
```

## Control flow
### Predict: Boundary case `heart_rate = 60`
**The trap:** many expect *bradycardia*, because 60 "feels low". But the condition is `< 60` (strictly less than), and 60 is **not** less than 60. It is also not `> 100`. So **both** specific branches are skipped and the `else` runs → **Normal**. This is the classic *off-by-one / boundary* bug: always check whether a threshold should be `<` or `<=`.

```python
heart_rate = 60
if heart_rate > 100:
    print('TACHYCARDIA')
elif heart_rate < 60:
    print('BRADYCARDIA')
else:
    print('Normal')
```

## While loops
### Predict: How many lines? (termination)
The condition is `index < len(durations) and durations[index] < 120`. Walk it through:

* `index = 0` → `60 < 120` is `True` → **print** Hernia repair, then `index = 1`.
* `index = 1` → `45 < 120` → **print** Appendicectomy, then `index = 2`.
* `index = 2` → `90 < 120` → **print** Cholecystectomy, then `index = 3`.
* `index = 3` → `180 < 120` is `False` → the `and` is `False` → **stop**.

**So it prints exactly 3 lines.** Mastectomy is never reached, and it would not print even if it were: `120 < 120` is `False`, the same boundary trap as `heart_rate = 60`. The loop terminates because each pass increments `index`, so eventually either a long case appears *or* `index` reaches the end of the list. **A common bug:** forgetting `index = index + 1` — then the condition never changes and the loop runs forever (an *infinite loop*). Also note Python *short-circuits* `and`: if the first part were false it would not even evaluate the second, which protects us from indexing past the end of the list.

```python
index = 0
while index < len(durations) and durations[index] < 120:
    print(procedures[index], durations[index], 'min')
    index = index + 1
```

## Capstone — who needs reviewing after surgery?
**Approach:** loop over the beds by index, so one index picks the matching value from each list. For each bed, score each observation with an `if` / `elif` / `else` chain (2 points first, then 1, otherwise 0), add the three, then choose the decision. The urgent count starts at 0 *before* the loop, so it is not reset for every bed.

**Expected output:** Beds 4 (colectomy) and 6 (laparotomy) are **urgent**; Beds 2 (appendicectomy) and 7 (hip replacement) need **review**; the rest are **routine**, apart from the stretch rule below. **2** beds need urgent review.

**Where it goes wrong:**

* **Bed 5** sits exactly on two boundaries: heart rate 130 and temperature 38.0. "Above 130" is `> 130`, so 130 scores 1, not 2, and 38.0 scores 0. With `>=` Bed 5 wrongly totals 4 and becomes **review**: the same boundary trap as `heart_rate = 60`.
* **Bed 7** only scores through the *low* thresholds (heart rate 48, respiratory rate 8, temperature 35.6), so a solution that checks only the high values misses it. Slow breathing, a slow heart and a low temperature after a hip replacement is the pattern of too much opioid: a high score is not only about fever and a fast heart.
* **Stretch:** **Bed 8** totals 2 (routine), but its respiratory rate of 25 scores 2 on its own, so the stretch rule raises it to **review**. Keeping each observation's points in its own variable makes that check easy.

```python
beds        = ['Bed 1', 'Bed 2', 'Bed 3', 'Bed 4', 'Bed 5', 'Bed 6', 'Bed 7', 'Bed 8']
operations  = ['Hernia repair', 'Appendicectomy', 'Mastectomy', 'Colectomy', 'Knee replacement', 'Laparotomy', 'Hip replacement', 'Cholecystectomy']
heart_rate  = [72, 118, 95, 134, 130, 125, 48, 88]
resp_rate   = [16, 24, 18, 28, 21, 27, 8, 25]
temperature = [36.8, 38.4, 37.1, 39.2, 38.0, 39.4, 35.6, 36.4]

urgent_count = 0
for index in range(len(beds)):
    hr = heart_rate[index]
    rr = resp_rate[index]
    temp = temperature[index]

    if hr > 130 or hr < 40:
        hr_points = 2
    elif hr > 110 or hr < 50:
        hr_points = 1
    else:
        hr_points = 0

    if rr > 24 or rr < 9:
        rr_points = 2
    elif rr > 20:
        rr_points = 1
    else:
        rr_points = 0

    if temp > 39.0 or temp < 35.0:
        temp_points = 2
    elif temp > 38.0 or temp < 36.0:
        temp_points = 1
    else:
        temp_points = 0

    score = hr_points + rr_points + temp_points

    if score >= 5:
        decision = 'urgent'
        urgent_count = urgent_count + 1
    elif score >= 3:
        decision = 'review'
    elif hr_points == 2 or rr_points == 2 or temp_points == 2:   # stretch
        decision = 'review'
    else:
        decision = 'routine'

    print(beds[index], operations[index], score, decision)

print('Beds needing urgent review:', urgent_count)
```

## Additional work 1 — BMI calculator
**Approach:** set the units' labels and formula choice with `if` / `else`, loop over a range of weights, classify each BMI with an `if` / `elif` chain, and print with commas between the values. `range` only yields whole numbers, so work out the 50% and 150% bounds and convert them with `int()` first.

```python
units = "metric"      # "metric" (kg, m) or "imperial" (lb, in)
weight = 100          # starting weight: kg if metric, lb if imperial
height = 1.8          # m if metric, in if imperial

if units == "metric":
    weight_unit = "kg"
    height_unit = "m"
else:
    weight_unit = "lb"
    height_unit = "in"

low = int(weight * 0.5)
high = int(weight * 1.5)

for w in range(low, high + 1, 10):
    if units == "metric":
        bmi = w / height ** 2
    else:
        bmi = 703 * w / height ** 2

    if bmi < 18.5:
        category = "underweight"
    elif bmi < 25:
        category = "healthy"
    elif bmi < 30:
        category = "overweight"
    elif bmi < 40:
        category = "obese"
    else:
        category = "severely obese"

    print(height, height_unit, w, weight_unit, round(bmi, 1), category)
```
